### Balancing the dataset

```python
from data_restructuration import get_transformed_data, get_data_loaders
```

```python
dataset = get_transformed_data(
    image_length=700, 
    image_heigth=700, 
    root=".\\MINI-DDSM-Complete-JPEG-82"
)
```

```python
train_loader, val_loader, test_loader = get_data_loaders(
    dataset=dataset,
    numero_random=42,
    batch_size=6,
    train_size=0.7,
    val_size=0.15,
    test_size=0.15
)
```

### Training the model

```python
import torchvision.models as models

from training_model import train_and_validate
```

```python
model = models.resnet18(weights=ResNet18_Weights.DEFAULT)

best_acc=0.0

train_and_validate(
    model=model, 
    train_loader=train_loader, 
    val_loader=val_loader, 
    criterion=torch.nn.CrossEntropyLoss(), 
    optimizer=torch.optim.Adam(model.parameters(), lr=0.0001),
    device=torch.device("cuda:0" if torch.cuda.is_available() else "cpu"),
    num_epochs=0,
    num_class=2,
    scheduler=torch.optim.lr_scheduler.ReduceLROnPlateau(optimizer, mode='max', factor=0.1, patience=3, threshold=0.0001)
)
```

### Evaluating the model

```python
predictions_np, labels_np, probabilities_np, accuracy = evaluate_model(
    path="best_resnet18_model_ordonnanceur.pth",
    test_loader=test_loader
)
print(f'Test_Accuracy : {accuracy*100:.4f}%')
```
